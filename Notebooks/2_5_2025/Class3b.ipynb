```python
import numpy as np
import matplotlib.pyplot as plt
```

We develop a simple code to solve the diffusion equation assuming a constant uplift term using an explicit formulation

```python
L = 1000
N = 101
K = 0.1
dx = L/(N-1)
dt = 200
U = 1e-4
x = np.linspace(0,L,N)
F = K*dt/dx**2
print('Courant=',F)
```

```python
h = np.zeros_like(x)
for step in range(100000):
    h[1:-1] = h[1:-1] + U*dt + F*(h[2:] - 2*h[1:-1] + h[:-2])
    if step%10000==0:
        plt.plot(x,h)
ha = -U/2/K*x**2 +U/2/K*L*x
plt.plot(x,ha,'-+')
```
